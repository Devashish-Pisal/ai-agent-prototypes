```python
import pandas as pd
from src.config import DATASET_FILE_PATH
from src.train_and_eval import (
    train_logistic_regression, train_random_forest, train_xgboost, evaluate_model, feature_importance,
    train_linear_regression, train_random_forest_regressor, evaluate_regression_model,
    evaluate_combined,
)
from src.data_processing import prepare_data
```

## 1.  Dataset Preparation

```python
df = pd.read_excel(DATASET_FILE_PATH)
(X_train, X_test, y_train, y_test,
 X_train_fee, X_test_fee, y_train_fee, y_test_fee,
 y_test_actual_commission) = prepare_data(df)

print("="*50)
print("CLASSIFICATION DATASET: ")
print(f"Train dataset: rows - {len(X_train)},  columns - {len(X_train.columns)}")
print(f"Test dataset: rows - {len(X_test)}, columns - {len(X_test.columns)}")
print("="*50)
print("REGRESSION DATASET: ")
print(f"Train dataset: rows - {len(X_train_fee)},  columns - {len(X_train_fee.columns)}")
print(f"Test dataset: rows - {len(X_test_fee)}, columns - {len(X_test_fee.columns)}")
print("="*50)
```

## 2. Classification

```python
print("CLASSIFICATION: will the request convert?")
print("="*50)

logistic_model = train_logistic_regression(X_train, y_train)
evaluate_model(logistic_model, X_test, y_test)
print("="*50)

random_forest_model = train_random_forest(X_train, y_train)
evaluate_model(random_forest_model, X_test, y_test)
feature_importance(random_forest_model, X_train.columns)
print("="*50)

xgboost_model = train_xgboost(X_train, y_train)
evaluate_model(xgboost_model, X_test, y_test)
feature_importance(xgboost_model, X_train.columns)
print("="*50)
```

## 3. Regression

```python
print("REGRESSION: how much commission if it converts?")
print("="*50)

linear_model = train_linear_regression(X_train_fee, y_train_fee)
evaluate_regression_model(linear_model, X_test_fee, y_test_fee)
print("="*50)

forest_regressor = train_random_forest_regressor(X_train_fee, y_train_fee)
evaluate_regression_model(forest_regressor, X_test_fee, y_test_fee)
print("="*50)
```

## 4. Combined

```python
print("COMBINED: expected commission per request")
print("="*50)
evaluate_combined(xgboost_model, forest_regressor, X_test, y_test_actual_commission)
print("="*50)
```
